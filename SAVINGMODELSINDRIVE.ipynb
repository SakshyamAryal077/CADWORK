```python
import os
import requests

PROJECT_DIR = "/content/drive/MyDrive/ECG_Project"
acs_dir = os.path.join(PROJECT_DIR, "ACS19955")
os.makedirs(acs_dir, exist_ok=True)

csv_zip = os.path.join(acs_dir, "CSV.zip")

url = "https://ndownloader.figshare.com/files/62951134"

if not os.path.exists(csv_zip):
    print("Downloading CSV...")
    r = requests.get(url, stream=True)
    r.raise_for_status()

    with open(csv_zip, "wb") as f:
        for chunk in r.iter_content(chunk_size=1024*1024):
            if chunk:
                f.write(chunk)

print("CSV downloaded.")
```

```python
row_zip = os.path.join(acs_dir, "ECG_row_data.zip")

url = "https://ndownloader.figshare.com/files/62951395"

if not os.path.exists(row_zip):
    print("Downloading ECG data...")
    r = requests.get(url, stream=True)
    r.raise_for_status()

    total = int(r.headers.get("content-length", 0))
    downloaded = 0

    with open(row_zip, "wb") as f:
        for chunk in r.iter_content(chunk_size=1024*1024):
            if chunk:
                f.write(chunk)
                downloaded += len(chunk)

                if total:
                    print(
                        f"\rDownloaded: {downloaded/1e9:.2f}/"
                        f"{total/1e9:.2f} GB",
                        end=""
                    )

print("\nECG data downloaded.")
```
